# Prefetch de Datasets (Plan A/Plan B)
## Master Oficial: Big Data Science

Este notebook se ejecuta antes de clase para dejar cacheados los datasets de Planetoid y evitar depender de internet durante la sesion.

### Objetivo
- Cachear Cora (Plan A) y CiteSeer (Plan B).
- Verificar que los datos quedan accesibles localmente en `../data/planetoid`.

### Convencion de ruta
- Asume que los notebooks viven en la carpeta `notebooks/`.
- Por eso usamos `DATA_DIR = Path("../data")` en estos materiales.

```python
from pathlib import Path

from torch_geometric.datasets import Planetoid

DATA_DIR = Path("../data")
PLANETOID_ROOT = DATA_DIR / "planetoid"
PLANETOID_ROOT.mkdir(parents=True, exist_ok=True)

print(f"Data dir: {DATA_DIR.resolve()}")
print(f"Planetoid root: {PLANETOID_ROOT.resolve()}")
```

```python
def prefetch_planetoid(name: str):
    print(f"\n=== Prefetch {name} ===")
    try:
        dataset = Planetoid(root=str(PLANETOID_ROOT), name=name)
        data = dataset[0]
        print(
            f"{name} cached OK | nodes={data.num_nodes} | edges={data.num_edges} | "
            f"features={dataset.num_features} | classes={dataset.num_classes}"
        )
        return {
            "dataset": name,
            "status": "OK",
            "nodes": int(data.num_nodes),
            "edges": int(data.num_edges),
            "features": int(dataset.num_features),
            "classes": int(dataset.num_classes),
        }
    except Exception as exc:
        print(f"{name} cache FAILED: {type(exc).__name__}: {exc}")
        return {
            "dataset": name,
            "status": "FAILED",
            "error": f"{type(exc).__name__}: {exc}",
        }
```

```python
results = [
    prefetch_planetoid("Cora"),
    prefetch_planetoid("CiteSeer"),
]

print("\n=== Summary ===")
for row in results:
    print(row)
```

### Criterio de salida
- Si Cora y CiteSeer aparecen con `status=OK`, la clase puede ejecutarse sin descargar datasets durante la demo.
- Si Cora falla pero CiteSeer esta OK, usar CiteSeer como fallback en el bonus final.
